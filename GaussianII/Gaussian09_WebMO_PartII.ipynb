```python
#@title Define callout functions (run this cell first) {display-mode: "form"}
from IPython.display import HTML, display

_question_index = 0  # Initialize global counter for questions
_indexed_questions = {}  # Stores {original_question_text: assigned_index}
_task_index = 0  # Initialize global counter for tasks
_indexed_tasks = {}  # Stores {original_task_text: assigned_index}

def callout(text, type="task", title=None):
    global _question_index
    global _indexed_questions
    global _task_index
    global _indexed_tasks

    #  This color palette was recommended by Claude to work well
    #  for both light and dark modes.
    colors = {
        "warning":   ("#D97706", "#1C1917", "⚠️ Warning"),
        "important": ("#DC2626", "#FFFFFF", "❕Important ❕"),
        "question":  ("#7C3AED", "#FFFFFF", "❔Question ❔"),
        "note":      ("#15803D", "#FFFFFF", "📘Note "),
        "task":      ("#2563EB", "#FFFFFF", "🎯 Task 🎯"),
        "tip":       ("#0D9488", "#1C1917", "💡Tip"),
    }

    bg, fg, default_label = colors.get(type, colors["task"])
    label = title if title else default_label

    if type == "question":
        original_text = text  # Store the original text before modification
        if original_text in _indexed_questions:
            # If this question has been seen before, use its assigned index
            assigned_idx = _indexed_questions[original_text]
            text = f"Q {assigned_idx}. {original_text}"
        else:
            # If it's a new question, increment and assign a new index
            _question_index += 1
            _indexed_questions[original_text] = _question_index
            text = f"Q {_question_index}. {original_text}"
    elif type == "task":
        original_text = text
        if original_text in _indexed_tasks:
            assigned_idx = _indexed_tasks[original_text]
            text = f"T {assigned_idx}. {original_text}"
        else:
            _task_index += 1
            _indexed_tasks[original_text] = _task_index
            text = f"T {_task_index}. {original_text}"

    display(HTML(f"""
<div style="background:{bg}; color:{fg}; padding:12px 16px;
border-radius:6px; margin:8px 0;">
<strong>{label}</strong>
<br>
<h3 style="color:{fg};">{text}</h3>
</div>
"""))
print("Callouts have been defined.")
```

The rotational constant is given by

$$\tilde{B} = \frac{h}{8\pi^2 c I}$$

where $c$ is the speed of light in a vacuum and $I$ is the moment of inertia,

$$I = \mu r^2, \qquad \mu = \frac{m_1 m_2}{m_1 + m_2},$$

with $r$ the bond length.

```python
#@title Question: Compute B̃ from r and isotope masses {display-mode: "form"}
callout("""Compute B̃ for HCl and DCl from your computed values of r and the isotope masses you find in the literature. How well do your values of B̃ compare with the computed values?""", type="question")
```

```python
#@title B̃ calculator (check your hand calculation)
#@markdown Enter isotope masses (in amu) from the literature and your computed bond lengths (in Å), then run the cell.
m_H = 0.0  #@param {type:"number"}
m_D = 0.0  #@param {type:"number"}
m_Cl = 0.0  #@param {type:"number"}
r_HCl = 0.0  #@param {type:"number"}
r_DCl = 0.0  #@param {type:"number"}

import math
h = 6.62607015e-34        # J s
c = 2.99792458e10         # cm/s
amu = 1.66053906660e-27   # kg

def B_tilde(m1, m2, r_angstrom):
    mu = m1 * m2 / (m1 + m2) * amu        # kg
    I = mu * (r_angstrom * 1e-10) ** 2    # kg m^2
    return h / (8 * math.pi ** 2 * c * I) # cm^-1

for name, m1, r in [("HCl", m_H, r_HCl), ("DCl", m_D, r_DCl)]:
    if m1 > 0 and m_Cl > 0 and r > 0:
        B = B_tilde(m1, m_Cl, r)
        print(f"{name}: B̃ = {B:.4f} cm^-1  (= {B * 29.9792458:.2f} GHz)")
    else:
        print(f"{name}: enter the masses and bond length above, then run the cell.")
```

### 3.4 C–C rotational barrier of ethane

#### 3.4.1 Introduction

In this section you will calculate the rotational barrier of the central C–C bond in ethane. Although "single" bonds are often said to exhibit "free rotation", there is in fact a small barrier to rotation. As the C–H bonds on either side of the C–C bond become "eclipsed" with bond rotation, the energy of the configuration rises. The height of the barrier is thought to be about 3 kcal/mol.

The calculation "drives" the dihedral angle through a series of values, holding the dihedral angle constant at each value while all other coordinates are optimized to minimize the energy.

The images below illustrate a dihedral angle of 0° between adjacent C–H bonds of ethane.
