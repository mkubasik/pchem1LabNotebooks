```python
#@title Define callout functions
from IPython.display import HTML, display

_question_index = 0 # Initialize global counter for questions
_indexed_questions = {} # Stores {original_question_text: assigned_index}
_task_index = 0 # Initialize global counter for tasks
_indexed_tasks = {} # Stores {original_task_text: assigned_index}

def callout(text, type="task", title=None):
    global _question_index
    global _indexed_questions
    global _task_index
    global _indexed_tasks

    colors = {
        "note":      ("#cce5ff", "#004085", "📘 Note"),
        "warning":   ("#fff3cd", "#856404", "⚠️ Warning"),
        "tip":       ("#d4edda", "#155724", "💡 Tip"),
        "important": ("#f8d7da", "#721c24", "❗ Important"),
        "question":  ("#e8d5f5", "#5b2c8d", "❓ Question"),
        "task":      ("#ffcc80", "#5b2c8d", "🔨 Task") # Deepened orange
    }
    bg, fg, default_label = colors.get(type, colors["task"])
    label = title if title else default_label

    if type == "question":
        original_text = text # Store the original text before modification
        if original_text in _indexed_questions:
            # If this question has been seen before, use its assigned index
            assigned_idx = _indexed_questions[original_text]
            text = f"Q {assigned_idx}. {original_text}"
        else:
            # If it's a new question, increment and assign a new index
            _question_index += 1
            _indexed_questions[original_text] = _question_index
            text = f"Q {_question_index}. {original_text}"
    elif type == "task":
        original_text = text
        if original_text in _indexed_tasks:
            assigned_idx = _indexed_tasks[original_text]
            text = f"T {assigned_idx}. {original_text}"
        else:
            _task_index += 1
            _indexed_tasks[original_text] = _task_index
            text = f"T {_task_index}. {original_text}"

    display(HTML(f"""
        <div style="background:{bg}; color:{fg}; padding:12px 16px;
                    border-radius:6px; margin:8px 0;">
            <strong>{label}</strong>
            <br>
            <h3>{text}</h3>
        </div>
    """))
print("Callouts have been defined.")

#possible palette:
#pink 2 F29191
#pink 1 F7ADAD
#Teal 2 B1E5E6
#Teal 1 CCFBFA

# Color palette from https://colorhunt.co/palettes/pastel
```

# Examples: Interactive Sin Waves

```python
#@title An interactive sin wave
def psi(n):

    psi_1 = []  ## This is an empty array.

    """Generate wavefunctions"""

    x_per_a = np.linspace(0, 1, 1001)
    for ii in np.arange(len(x_per_a)): psi_1.append(np.sin(np.pi * n * x_per_a[ii]))

    plt.figure(figsize=(4,3))
    plt.xlabel('(x)', fontsize=14)
    plt.title(rf"Sin(n $\pi$ x), n = {n}")
    #plt.ylabel('f(v)', fontsize=16)

    plt.plot(x_per_a, psi_1, alpha=0.9, color='red',  label=r'Sin(n $\pi$ x)')
    plt.legend(loc='lower left')

    plt.grid()

interactive_plot = interactive(psi,n=widgets.IntSlider(min=1, max=10, step=1, value=1))
interactive_plot
```

```python
#@title Two interactive sin waves
def psi(n):

    psi_1, psi_2 = [], []

    """Generate wavefunctions"""

    x_per_a = np.linspace(0, 1, 1001)
    for ii in np.arange(len(x_per_a)): psi_1.append(np.sin(np.pi * n * x_per_a[ii]))
    for ii in np.arange(len(x_per_a)): psi_2.append(np.sin(np.pi * (n+1) * x_per_a[ii]))


    plt.figure(figsize=(7,7))
    plt.xlabel('(x)', fontsize=14)
    plt.title(rf"Sin(n $\pi$ x) and Sin((n+1) $\pi$ x), n = {n}")
    #plt.ylabel('f(v)', fontsize=16)

    plt.plot(x_per_a, psi_1, alpha=0.9, color='red',  label=r'Sin(n $\pi$ x)')
    plt.plot(x_per_a, psi_2, alpha=0.9, color='black',  label=r'Sin((n+1) $\pi$ x)')
    plt.legend(loc='lower left')

    plt.grid()

interactive_plot = interactive(psi,n=widgets.IntSlider(min=1, max=10, step=1, value=1))
interactive_plot
```

```python
#@title Example plot: $Sin(n\pi x)^2$


def psi(n=1):

    psi_1, psi_2 = [], []

    """Generate wavefunctions"""

    x_per_a = np.linspace(0, 1, 1001)
    psi_1 = np.sqrt(2) * np.sin(np.pi * n * x_per_a)


    plt.figure(figsize=(6, 6))
    plt.xlabel('x/a', fontsize=14)
    plt.ylabel(r'$\psi^2$', fontsize=16)
    plt.grid()

    plt.plot(x_per_a, psi_1**2, alpha=0.9, color='red',  label='first')

    plt.fill_between(x_per_a, psi_1**2, alpha=0.1, color='red')
    plt.title(r"$\sqrt{2}$Sin($n \pi x$)$^2$")
    plt.xticks(np.arange(0, 1.1, 0.1)) # Add tick marks every 0.1 increment
interactive_plot = interactive(psi,n = widgets.IntSlider(min=1, max=10, step=1, value=1))

interactive_plot
```

# Example.  Bokeh

```python
callout("The most common plotting library in Python \
        is matplotlib, used above.  However, 'most common' and 'best' \
        are not always the same.  The remainder of the notebook employs \
        the powerful visualization library 'bokeh'.",
        "note")
```

```python
#@title A bokeh example.  Randomly positioned circles of random radii
N = 4096                           # Number of points
x = np.random.random(size=N) * 100 # Generates N points between 0 and 100
y = np.random.random(size=N) * 100 # Generates N points between 0 and 100
radii = np.random.random(size=N) * 1.5 # Generates N radii between 0 and 1

# Generate, from the x and y arrays, red and green color scales for RGB.
colors = ["#%02x%02x%02x" % (r, g, 150) for r, g in zip(np.floor(50+2*x).astype(int), np.floor(30+2*y).astype(int))]

# Establish a "figure object" and give it the name "p"
p = figure()
#
p.circle(x,                # x data, 0-100, randomly
         y,                # y data, 0-100, randomly
         radius=radii,     # radii is an array of random radii
         fill_color=colors, # colors is an array of lenght N with differen RGB settings
         fill_alpha=0.6,    # Transparency setting
         line_color=None   # Do not "connect the dots."
         )
p.title.text = "Random Data: Cirlces of random position with randomly sized radii \nColors are scaled with increading green in y and increasing red in x"
show(p)
```

```python
#@title The First Few Vibrational wavefunctions

from bokeh.models import Label, Range1d, Title


p = figure(width=650,
           height=600,
           x_range=Range1d(-6, 6),
           y_range=Range1d(0, 8),
           toolbar_location="right")
p.xaxis.axis_label = r"$$q$$"
p.xaxis.axis_label_text_font_size = "14pt"
p.yaxis.visible = False
p.xgrid.visible = False
p.ygrid.visible = False

title = [
    r"$$\text{Wavefunction } \psi_v(q) \text{ of first 8 mode solutions of Schrodinger's equation }" +
        r" -\frac{1}{2}\frac{d^2\psi}{dq^2} + \frac{1}{2}q^2\psi = \frac{E}{\hbar\omega}\psi$$",
    r"$$\text{Each wavefunction is labelled with its quantum number, } v \text{, and energy } E_v$$",
    r"$$\text{in a potential } V(q) = \frac{q^2}{2} \text{ shown by the dashed line.}$$",
]
for text in reversed(title):
    p.add_layout(Title(text=text, text_font_style="normal"), "above")

q = np.linspace(-6, 6, 100)
yscale = 0.75
number_of_modes = 8

for v in range(number_of_modes):
    H_v = hermite(v)
    N_v = (np.pi**0.5 * 2**v * factorial(v))**(-0.5)
    psi = N_v*H_v(q)*np.exp(-q**2/2)
    E_v = v + 0.5  # Use energy level as y-offset.

    y = yscale*psi + E_v
    yupper = np.where(y >= E_v, y, E_v)
    ylower = np.where(y <= E_v, y, E_v)

    p.varea(q, yupper, E_v, fill_color="coral")
    p.varea(q, ylower, E_v, fill_color="orange")
    p.line(q, y, color="red", line_width=2)

    p.add_layout(Label(x=-5.8,
                       y=E_v,
                       y_offset=-21,
                       text=r"$$v = " + str(v) + r"$$",
                       text_font_size='12pt'))
    p.add_layout(Label(x=3.9,
                       y=E_v,
                       y_offset=-20,
                       text=r"$$E_" + str(v) + r" = (" + str(2*v+1) + r"/2) \hbar\omega$$",
                       text_font_size='10pt'
                       ))

V = q**2 / 2
p.line(q,                   # Add the quadratic potential to the plot
       V,
       line_color="black",
       line_width=2,
       line_dash="dashed")

show(p)
```

```python
#@title Some more code
import numpy as np
from bokeh.plotting import figure, show
from bokeh.io import output_notebook, curdoc
from bokeh.models import ColumnDataSource

# Ensure Bokeh is configured for notebook output if not already
output_notebook()

# Attempt to clear the current Bokeh document to remove lingering models
curdoc().clear()

p1 = figure(title="Molar Concentration vs Time of A, B, and C",
           x_range=(0, time_seconds.max()),
           y_range=(0, max(a_conc.max(), b_conc.max(), c_conc.max()) * 1.1) # Add some padding
          )
# Clear any potentially lingering renderers from previous executions on this specific figure
p1.renderers = []
p1.xaxis.axis_label = "time(s)"
p1.yaxis.axis_label = "Concentration (mol/L)"

# Create a ColumnDataSource for all data to be plotted
source = ColumnDataSource(data=dict(
    time=time_seconds,
    a=a_conc,
    b=b_conc,
    c=c_conc
))

# Add a line for each concentration to connect the points
p1.line(x='time', y='a', source=source, line_color='red', legend_label='A_conc', line_width=2)
p1.line(x='time', y='b', source=source, line_color='blue', legend_label='B_conc', line_width=2)
p1.line(x='time', y='c', source=source, line_color='green', legend_label='C_conc', line_width=2)

p1.legend.location = "top_left"
p1.grid.grid_line_alpha = 0.5

show(p1)
```
