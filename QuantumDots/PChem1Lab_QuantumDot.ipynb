```python
E_nano = u.planck_constant * u.speed_of_light / (700 * u.nm)
```

```python
E_bulk = u.planck_constant * u.speed_of_light / (810 * u.nm)
```

```python
epsilon = 11.0
me_star = 0.16
mh_star = 1.3
```

```python
a = E_nano - E_bulk
b = 1.8 * (u.e)**2 / (4 * u.pi * epsilon * u.eps_0)
c = -1.0 * (u.planck_constant)**2 / ( 8 * u.m_e) * (1/me_star + 1/mh_star)
```

```python
r = (-b + np.sqrt(b**2 - 4 * a * c))/ (2 * a)
r.to_base_units()
```
